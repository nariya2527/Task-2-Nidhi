```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_csv("cleaned_fake_job.csv")
```

```python
# Check Number of Rows and columns
print("Dataset Shape:",df.shape)
```

```python
# Displaying First 5 rows
df.head(5)
```

```python
# Understanding Dataset columns and datatypes
df.info()
```

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Calculating percentage of missing values
missing_percentage = (df.isnull().sum() / len(df))*100
print(missing_percentage)
```

```python
# Removing columns which are not very useful
df.drop(columns=['department','benefits'],inplace=True)
print("Unnecessary Columns Removed")
```

```python
# Fill missing values in important columns
df['company_profile'] = df['company_profile'].fillna("Not Available")
df['description'] = df['description'].fillna("No Desciption")
df['requirements'] = df['requirements'].fillna("Requirements Not Mentioned")

print("Missing Values Handled Succesfully")
```

```python
# Remove Duplicate rows
before = df.shape[0]
df.drop_duplicates(inplace=True)
after = df.shape[0]

print("Duplicate Rows removed:",before - after)
```

```python
# Counting Fake and Real Jobs
df['fraudulent'].value_counts()
```

```python
# Import matplotlib
import matplotlib.pyplot as plt
```

```python
# Plotting fake vs real jobs
df['fraudulent'].value_counts().plot(kind='bar')
plt.title("Fake vs Real Jobs Posting")
plt.xlabel("0=Real Jobs,1 = Fake Jobs")
plt.ylabel("Count")
plt.show()
```

```python
# Check most common employment types
df['employment_type'].value_counts()
```

```python
# Top job locations
df['location'].value_counts().head()
```

```python
# Saving cleaned dataset
df.to_csv('cleaned_fake_jobs.csv',index = False)
print("Cleaned Dataset Saved Successfully")
```
